# Pretrained Model



```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array  # alternative to PIL
from tensorflow.keras.applications.vgg16 import preprocess_input, VGG16, decode_predictions
print("Done")
```

```python
import matplotlib.pyplot as plt

image = load_img("oldtimer.jpg", target_size=(224, 224))
image = img_to_array(image)
print(image.shape) # Single Picture . int values from 0 to 255
```

```python
print(image[0][0:10])
```

```python
image = image.reshape((1, 224, 224, 3)) # makes one image to a list of one image
#plt.imshow(image[0]/255)
image = preprocess_input(image)
image.shape
plt.imshow(image[0])  ## this is how the model gets the image
```

```python
print(image[0][0][0:10])
```

```python
print("Instantiating model..")
model = VGG16()
model.summary()
```

*Dokumentation:*

https://www.tensorflow.org/api_docs/python/tf/keras/applications/VGG16

```python
print("Inference..")
pred = model.predict(image)
print(f"Result: {pred.shape}")
```

```python
# looking at the content
pred
```

```python
import numpy as np
np.argmax(pred, axis=1)
```

```python
pred[0][751]
```

```python
decoded_predictions = decode_predictions(pred, top=5)
decoded_predictions
```
