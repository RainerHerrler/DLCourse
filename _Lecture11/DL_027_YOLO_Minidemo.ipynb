```python
#!pip install ultralytics
from ultralytics import YOLO
import matplotlib.pyplot as plt

model = YOLO("yolo11n.pt")    # small pretrained model (COCO, 80 classes)
#model = YOLO("yolo26n.pt")   # YOLO26 – latest model from ultralytics

results = model("RoboterLab.png")
#results[0].show()            # boxes + classes + confidences

img = results[0].plot()       # returns a NumPy array (BGR) with drawn boxes
plt.figure(figsize=(10, 8))
plt.imshow(img[..., ::-1])    # convert BGR -> RGB
plt.axis("off")
plt.show()

```

```python
model = YOLO("yolov8x-worldv2.pt")
model.set_classes(["robot", "Soldering station", "spool", "sinus" ,"planted pot", "Screw", "Book", "Sketch", "Robot", "Head", "Cup", "Drawer", "yellow paper"])
results = model("RoboterLab.png")
```
